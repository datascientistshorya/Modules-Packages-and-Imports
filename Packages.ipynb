## 2. Validation Rules

Before calculating sales metrics, apply these rules:

1. A sales record must be a dictionary.
2. The `amount` field must exist and cannot be `None`.
3. The amount must be numeric. Numeric strings such as `"1200"` are not accepted.
4. Boolean values (`True` and `False`) are not accepted as sales amounts.
5. Negative sales amounts are invalid.
6. Zero is a valid sales amount.
7. Invalid or excluded records are not included in revenue, average order value, highest-value order, or customer order counts.
8. Customer counts include only customers with valid sales records.

For this dataset, the record with a missing amount and the record with a negative amount will be excluded.

## 3. `analytics/cleaning.py`

This module removes missing records and validates sales amounts.

```python
from numbers import Real


def remove_missing_sales(orders):
    """Remove records with missing sales amounts."""
    cleaned_orders = []

    for order in orders:
        if not isinstance(order, dict):
            continue

        if order.get("amount") is None:
            continue

        cleaned_orders.append(order)

    return cleaned_orders


def validate_sales(orders):
    """Keep records with numeric, non-negative sales amounts."""
    valid_orders = []

    for order in orders:
        if not isinstance(order, dict):
            continue

        amount = order.get("amount")

        # Reject missing amounts and Boolean values
        if amount is None or isinstance(amount, bool):
            continue

        # Reject non-numeric amounts
        if not isinstance(amount, Real):
            continue

        # Reject negative amounts
        if amount < 0:
            continue

        valid_orders.append(order)

    return valid_orders
```

## 4. `analytics/sales.py`

This module calculates revenue metrics from valid orders.

```python
def total_revenue(orders):
    return sum(order["amount"] for order in orders)


def average_order_value(orders):
    if not orders:
        return 0

    return total_revenue(orders) / len(orders)


def highest_value_order(orders):
    if not orders:
        return None

    return max(orders, key=lambda order: order["amount"])
```

## 5. `analytics/customers.py`

This module analyzes unique customers and identifies the customer with the most valid orders.

```python
from collections import Counter


def unique_customer_count(orders):
    customer_ids = {
        order["customer_id"]
        for order in orders
        if order.get("customer_id") is not None
    }

    return len(customer_ids)


def customer_with_most_orders(orders):
    customer_ids = [
        order["customer_id"]
        for order in orders
        if order.get("customer_id") is not None
    ]

    if not customer_ids:
        return None

    counts = Counter(customer_ids)

    return counts.most_common(1)[0][0]
```

If two customers tie for the highest number of valid orders, this function returns the first one encountered in the data.

## 6. `analytics/reporting.py`

This module imports the functions from the other modules, processes the records, and generates a readable report.

```python
from analytics.cleaning import remove_missing_sales, validate_sales
from analytics.sales import (
    total_revenue,
    average_order_value,
    highest_value_order
)
from analytics.customers import (
    unique_customer_count,
    customer_with_most_orders
)


def generate_sales_report(orders):
    # Remove records with missing amounts
    non_missing_orders = remove_missing_sales(orders)

    # Validate the remaining records
    valid_orders = validate_sales(non_missing_orders)

    # Calculate the number of excluded records
    excluded_orders = len(orders) - len(valid_orders)

    # Calculate sales metrics
    revenue = total_revenue(valid_orders)
    average_value = average_order_value(valid_orders)
    highest_order = highest_value_order(valid_orders)

    # Calculate customer metrics
    unique_customers = unique_customer_count(valid_orders)
    top_customer = customer_with_most_orders(valid_orders)

    # Build the report
    report = [
        "========== SALES ANALYTICS REPORT ==========",
        f"Total input orders: {len(orders)}",
        f"Valid orders: {len(valid_orders)}",
        f"Invalid or excluded orders: {excluded_orders}",
        "",
        "SALES METRICS",
        f"Total revenue: {revenue:.2f}",
        f"Average order value: {average_value:.2f}",
    ]

    if highest_order is not None:
        report.append(
            f"Highest-value order: Order {highest_order['order_id']} "
            f"(Amount: {highest_order['amount']:.2f})"
        )
    else:
        report.append("Highest-value order: None")

    report.extend([
        "",
        "CUSTOMER METRICS",
        f"Unique customers: {unique_customers}",
        f"Customer with most valid orders: {top_customer}",
        "============================================"
    ])

    return "\n".join(report)
```
