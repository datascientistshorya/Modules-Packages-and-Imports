#### Q4 Import multiple modules

#### Q6 Understand module contents
Import math and complete these tasks:
1. Print the value of math.pi.

3. Check whether math has a sqrt attribute using hasattr().
4. Display the names available in the module using dir(math).
Requirement: Explain the difference between a built-in function and a module function.

```python
import math as m
print(m.pi)
print(abs(-78.56))
print(hasattr(m,'sqrt'))
print(dir(m))
#A built-in function is available directly in Python without importing anything, such as print(), len(), and sum(). A module function belongs 
#to a specific module and usually requires importing that module, such as math.sqrt() or random.randint().
```
