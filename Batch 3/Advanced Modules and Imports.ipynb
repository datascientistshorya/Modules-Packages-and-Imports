# Q3. Create a Plugin-Based Text Analyzer

## 1. Project Structure

Create the following folders and files:

```text
text_analyzer/
│
├── main.py
│
└── analyzers/
    ├── __init__.py
    ├── word_count.py
    ├── character_count.py
    └── sentence_count.py
```

## 2. `analyzers/word_count.py`

Create a function that counts the words in a text.

```python
def count_words(text):
    return len(text.split())
```

## 3. `analyzers/character_count.py`

Create a function that counts characters while excluding spaces.

```python
def count_characters(text):
    return len(text.replace(" ", ""))
```

## 4. `analyzers/sentence_count.py`

Create a function that counts sentences ending with a period, exclamation mark, or question mark.

```python
import re


def count_sentences(text):
    sentences = re.findall(r'[^.!?]+[.!?]+', text)
    return len(sentences)
```

## 5. `analyzers/__init__.py`

Leave this file empty. It marks `analyzers` as a regular Python package.

## 6. `main.py`

Import all three functions, analyze the given text, and print a report.

```python
from analyzers.word_count import count_words
from analyzers.character_count import count_characters
from analyzers.sentence_count import count_sentences

# Sample text
text = """Python makes data analysis easier.Reusable modules keep code organized.Good structure improves maintainability!"""

# Analyze the text
words = count_words(text)
characters = count_characters(text)
sentences = count_sentences(text)

# Print the report
print("===== TEXT ANALYSIS REPORT =====")
print("Word count:", words)
print("Character count (excluding spaces):", characters)
print("Sentence count:", sentences)
print("================================")
```

## 7. Expected Output

```text
===== TEXT ANALYSIS REPORT =====
Word count: 13
Character count (excluding spaces): 99
Sentence count: 3
================================
```

**Note:** The character count excludes ordinary spaces but includes punctuation. The word count uses whitespace to separate words, so it treats the text `easier.Reusable` as one word because there is no space after the period. If you want punctuation-aware word counting, you can improve the function later.

## 8. How to Add a Fourth Analyzer

Suppose you want to add an analyzer that counts vowels.

**Step 1:** Create a new file named `vowel_count.py` inside the `analyzers` folder.

```python
def count_vowels(text):
    vowels = "aeiouAEIOU"
    return sum(1 for char in text if char in vowels)
```

**Step 2:** Import the new function into `main.py`.

```python
from analyzers.vowel_count import count_vowels
```

**Step 3:** Call the function and print its result.

```python
print("Vowel count:", count_vowels(text))
```

You do not need to rewrite the existing word, character, or sentence counting functions. You simply add a new module and import its function wherever you need it.

## 9. Key Concepts Learned

- **Modules:** Each analyzer is stored in a separate Python file.
- **Packages:** The `analyzers` folder groups related modules.
- **Imports:** `main.py` reuses functions from different modules.
- **Modularity:** Each function has one clear responsibility.
- **Extensibility:** New analyzers can be added without changing existing analyzer functions.
